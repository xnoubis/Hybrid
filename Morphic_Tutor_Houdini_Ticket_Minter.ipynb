```

"""
Morphic Tutor — Domain-Crossing Credential Minter

This module implements the "Houdini Ticket" architecture, transforming
cognitive traversals into portable, API-agnostic, multi-strata artifacts.

Format: [ Handle (Sigil) | World Hash (Terrain) | Proof Receipt (Attestation) ]
"""

import hashlib
import json
import secrets
from dataclasses import dataclass, field, asdict
from datetime import datetime, timezone
from typing import Optional, Dict, List, Any

# The classes StereogramToken and Terrain are expected to be defined in preceding cells
# from token import StereogramToken
# from terrain import Terrain


# ═══════════════════════════════════════════════════════════
# MULTI-STRATA ENGINE
# ═══════════════════════════════════════════════════════════

@dataclass
class CognitiveStrata:
    """
    Objective Humility Embedded:
    Forcing the separation of facts, narratives, myths, and operations.
    Collapsing these layers is forbidden.
    """
    observations: List[str] = field(default_factory=list)    # Descriptive
    interpretations: List[str] = field(default_factory=list) # Narrative
    projections: List[str] = field(default_factory=list)     # Mythic/Imaginal
    actions: List[str] = field(default_factory=list)         # Operational

    def to_dict(self) -> dict:
        return asdict(self)


# ═══════════════════════════════════════════════════════════
# THE CORE HEURISTIC
# ═══════════════════════════════════════════════════════════

@dataclass
class MorphicState:
    """The Persistent Cognitive Spine."""
    declared_invariants: List[str]
    implied_invariants: List[str]
    open_questions: List[str]
    risk_flags: List[str]
    strata: CognitiveStrata
    drift_score: float
    branch_parent: Optional[str] = None

    def to_dict(self) -> dict:
        return {
            "declared_invariants": self.declared_invariants,
            "implied_invariants": self.implied_invariants,
            "open_questions": self.open_questions,
            "risk_flags": self.risk_flags,
            "strata_breakdown": self.strata.to_dict(),
            "drift_score": round(self.drift_score, 4),
            "branch_parent": self.branch_parent,
        }


# ═══════════════════════════════════════════════════════════
# HOUDINI TICKET FORMAT
# ═══════════════════════════════════════════════════════════

@dataclass
class ProofReceipt:
    """The Attestation layer of the Houdini Ticket."""
    morphic_state: dict
    stereogram_operator_hash: str
    snap_delta_entropy: float
    timestamp: str
    handle_salt: str = "" # Added to store the salt used for handle generation

@dataclass
class HoudiniTicket:
    """
    The ultimate portable artifact.
    Can be carried across platforms, UI sandboxes, and models.
    """
    handle: str            # The Sigil (ID)
    world_hash: str        # The Terrain State
    proof_receipt: dict    # The Attestation

    def to_json(self) -> str:
        return json.dumps(asdict(self), indent=2, sort_keys=True)


# ═══════════════════════════════════════════════════════════
# THE MINTER
# ═══════════════════════════════════════════════════════════

class MorphicTutorMinter:
    """
    Mints Houdini Tickets.
    Does not nag. Does not collapse strata. Just freezes drift as an artifact.
    """

    def synthesize_sigil(
        self,
        state: MorphicState,
        terrain: Terrain,
        token: StereogramToken,
        salt: Optional[str] = None # Added to allow passing a specific salt
    ) -> str:
        """
        Synthesizes a unique 'Handle' (Sigil) for the artifact.

        This acts as the ID for the Persistent Cognitive Spine. It cryptographically
        binds the human's subjective state (invariants, strata) to the machine's
        objective state (terrain topology, token operator).
        """
        if salt is None:
            salt = secrets.token_hex(8) # Generate new salt if not provided

        payload = {
            "morphic_state": state.to_dict(),
            "terrain_hash": terrain.to_dict(), # Incorporates topology/feature distribution
            "token_operator": token.operator_hash,
            "salt": salt                       # Use provided or newly generated salt
        }

        payload_bytes = json.dumps(payload, sort_keys=True).encode('utf-8')

        # The Sigil: A localized, URL-safe, domain-crossing identifier
        sigil = "sigil_" + hashlib.sha256(payload_bytes).hexdigest()[:24]
        return sigil


    def export(
        self,
        state: MorphicState,
        terrain: Terrain,
        token: StereogramToken
    ) -> HoudiniTicket:
        """
        Generates the exportable Houdini Ticket.

        This artifact is sovereign: API-agnostic, portable, and immediately
        composable by any external Nintendo-principle UI layer.
        """
        # Generate salt for handle and store it in the receipt
        handle_generation_salt = secrets.token_hex(8)

        # 1. Synthesize the Handle using the generated salt
        handle = self.synthesize_sigil(state, terrain, token, salt=handle_generation_salt)

        # 2. Compute the World Hash (Terrain State)
        # Represents the latent topology at the exact moment of the snap
        world_payload = json.dumps(terrain.to_dict(), sort_keys=True).encode('utf-8')
        world_hash = "world_" + hashlib.sha256(world_payload).hexdigest()[:16]

        # 3. Compile the Proof Receipt
        snap_delta = token.snap_event.delta_entropy if token.snap_event else 0.0

        receipt = ProofReceipt(
            morphic_state=state.to_dict(),
            stereogram_operator_hash=token.operator_hash,
            snap_delta_entropy=snap_delta,
            timestamp=datetime.now(timezone.utc).isoformat(),
            handle_salt=handle_generation_salt # Store the salt in the receipt
        )

        # 4. Mint the Ticket
        ticket = HoudiniTicket(
            handle=handle,
            world_hash=world_hash,
            proof_receipt=asdict(receipt)
        )

        # (Optional) Trigger a precision interruption if drift is too high or invariants are missing
        self._check_cognitive_hygiene(state)

        return ticket


    def _check_cognitive_hygiene(self, state: MorphicState) -> None:
        """
        The 'Precision Interruption' mechanism.
        Does not stop the export, but flags the structural integrity.
        """
        if state.drift_score > 0.8:
            # High drift detected. The UI layer should visually prompt branching.
            state.risk_flags.append("CRITICAL_DRIFT: Improvisation outpacing resolution.")

        if not state.declared_invariants and not state.implied_invariants:
            # Missing invariants. The UI should gently interrupt.
            state.risk_flags.append("COLLAPSE_RISK: No structural invariants anchored.")
```

```
import hashlib
import json

def validate_houdini_ticket(
    ticket: HoudiniTicket,
    original_state: MorphicState,
    original_terrain: Terrain,
    original_token: StereogramToken,
    minter_instance: MorphicTutorMinter
) -> bool:
    """
    Validates a HoudiniTicket by re-computing its handle and world_hash
    from the original components and comparing them to the ticket's values.
    """
    print("--- Starting HoudiniTicket Validation ---")

    # 1. Re-synthesize the handle, using the salt stored in the ticket's proof_receipt
    stored_handle_salt = ticket.proof_receipt.get('handle_salt')
    if not stored_handle_salt:
        print("Error: No handle_salt found in ProofReceipt. Cannot deterministically validate handle.")
        return False

    recomputed_handle = minter_instance.synthesize_sigil(
        original_state,
        original_terrain,
        original_token,
        salt=stored_handle_salt # Pass the stored salt for re-computation
    )
    handle_match = (recomputed_handle == ticket.handle)
    print(f"Original Handle: {ticket.handle}")
    print(f"Recomputed Handle: {recomputed_handle}")
    print(f"Handle Match: {handle_match}")

    # 2. Re-compute the world hash
    world_payload = json.dumps(original_terrain.to_dict(), sort_keys=True).encode('utf-8')
    recomputed_world_hash = "world_" + hashlib.sha256(world_payload).hexdigest()[:16]
    world_hash_match = (recomputed_world_hash == ticket.world_hash)
    print(f"Original World Hash: {ticket.world_hash}")
    print(f"Recomputed World Hash: {recomputed_world_hash}")
    print(f"World Hash Match: {world_hash_match}")

    # 3. Validate proof_receipt content against original components
    # This part assumes that the .to_dict() representations are stable
    morphic_state_in_receipt = ticket.proof_receipt['morphic_state']
    original_morphic_state_dict = original_state.to_dict()
    state_receipt_match = (morphic_state_in_receipt == original_morphic_state_dict)
    print(f"Morphic State in Receipt Matches Original: {state_receipt_match}")

    stereogram_operator_in_receipt = ticket.proof_receipt['stereogram_operator_hash']
    token_receipt_match = (stereogram_operator_in_receipt == original_token.operator_hash)
    print(f"Stereogram Operator in Receipt Matches Original: {token_receipt_match}")

    # Note: timestamp and snap_delta_entropy are part of the receipt but not directly re-computed from inputs
    # The primary focus for this validation is handle, world_hash, and the consistency of the stored state/token hashes.

    is_valid = handle_match and world_hash_match and state_receipt_match and token_receipt_match
    print(f"--- HoudiniTicket is Valid: {is_valid} ---")

    return is_valid

# Now, call the validation function with the existing instances
is_ticket_valid = validate_houdini_ticket(
    ticket=houdini_ticket,
    original_state=morphic_state_object,
    original_terrain=terrain_instance,
    original_token=stereogram_token_instance,
    minter_instance=minter
)

if is_ticket_valid:
    print("Validation successful! The HoudiniTicket is internally consistent.")
else:
    print("Validation failed! The HoudiniTicket has inconsistencies.")
```

With the `HoudiniTicket` now re-exported to include the `handle_salt` in its `proof_receipt`, the final step is to re-run the `validate_houdini_ticket` function to confirm that the validation passes successfully, as required by the main task.



```
import hashlib
import json

def validate_houdini_ticket(
    ticket: HoudiniTicket,
    original_state: MorphicState,
    original_terrain: Terrain,
    original_token: StereogramToken,
    minter_instance: MorphicTutorMinter
) -> bool:
    """
    Validates a HoudiniTicket by re-computing its handle and world_hash
    from the original components and comparing them to the ticket's values.
    """
    print("--- Starting HoudiniTicket Validation ---")

    # 1. Re-synthesize the handle, using the salt stored in the ticket's proof_receipt
    stored_handle_salt = ticket.proof_receipt.get('handle_salt')
    if not stored_handle_salt:
        print("Error: No handle_salt found in ProofReceipt. Cannot deterministically validate handle.")
        return False

    recomputed_handle = minter_instance.synthesize_sigil(
        original_state,
        original_terrain,
        original_token,
        salt=stored_handle_salt # Pass the stored salt for re-computation
    )
    handle_match = (recomputed_handle == ticket.handle)
    print(f"Original Handle: {ticket.handle}")
    print(f"Recomputed Handle: {recomputed_handle}")
    print(f"Handle Match: {handle_match}")

    # 2. Re-compute the world hash
    world_payload = json.dumps(original_terrain.to_dict(), sort_keys=True).encode('utf-8')
    recomputed_world_hash = "world_" + hashlib.sha256(world_payload).hexdigest()[:16]
    world_hash_match = (recomputed_world_hash == ticket.world_hash)
    print(f"Original World Hash: {ticket.world_hash}")
    print(f"Recomputed World Hash: {recomputed_world_hash}")
    print(f"World Hash Match: {world_hash_match}")

    # 3. Validate proof_receipt content against original components
    # This part assumes that the .to_dict() representations are stable
    morphic_state_in_receipt = ticket.proof_receipt['morphic_state']
    original_morphic_state_dict = original_state.to_dict()
    state_receipt_match = (morphic_state_in_receipt == original_morphic_state_dict)
    print(f"Morphic State in Receipt Matches Original: {state_receipt_match}")

    stereogram_operator_in_receipt = ticket.proof_receipt['stereogram_operator_hash']
    token_receipt_match = (stereogram_operator_in_receipt == original_token.operator_hash)
    print(f"Stereogram Operator in Receipt Matches Original: {token_receipt_match}")

    # Note: timestamp and snap_delta_entropy are part of the receipt but not directly re-computed from inputs
    # The primary focus for this validation is handle, world_hash, and the consistency of the stored state/token hashes.

    is_valid = handle_match and world_hash_match and state_receipt_match and token_receipt_match
    print(f"--- HoudiniTicket is Valid: {is_valid} ---")

    return is_valid

# Now, call the validation function with the existing instances
is_ticket_valid = validate_houdini_ticket(
    ticket=houdini_ticket,
    original_state=morphic_state_object,
    original_terrain=terrain_instance,
    original_token=stereogram_token_instance,
    minter_instance=minter
)

if is_ticket_valid:
    print("Validation successful! The HoudiniTicket is internally consistent.")
else:
    print("Validation failed! The HoudiniTicket has inconsistencies.")
```
